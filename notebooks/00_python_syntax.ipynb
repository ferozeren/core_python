 # Python Syntax

## Precedence ()


Python Operator Precedence Table

| Precedence | Operator(s) | Description | Associativity |
| --- | --- | --- | --- |
| **1 (Highest)** | `()` | Parentheses (grouping), function calls, slicing, subscription | Left-to-right |
| **2** | `**` | Exponentiation (power) | **Right-to-left** |
| **3** | `+x`, `-x`, `~x` | Unary plus, Unary minus, Bitwise NOT | Left-to-right |
| **4** | `*`, `/`, `//`, `%` | Multiplication, Division, Floor division, Modulus | Left-to-right |
| **5** | `+`, `-` | Addition, Subtraction | Left-to-right |
| **6** | `<<`, `>>` | Bitwise left and right shifts | Left-to-right |
| **7** | `&` | Bitwise AND | Left-to-right |
| **8** | `^` | Bitwise XOR | Left-to-right |
| **9** | `\\|` | Bitwise OR | Left-to-right |
| **10** | `==`, `!=`, `>`, `>=`, `<`, `<=`, `is`, `is not`, `in`, `not in` | Comparisons, Identity, and Membership | Non-associative (Chained) |
| **11** | `not` | Logical NOT | Left-to-right |
| **12** | `and` | Logical AND | Left-to-right |
| **13** | `or` | Logical OR | Left-to-right |
| **14** | `if - else` | Conditional expression (Ternary operator) | Right-to-left |
| **15** | `lambda` | Lambda expressions | \\- |
| **16** | `:=` | Assignment expression (Walrus operator) | \\- |
| **17 (Lowest)** | `=`, `+=`, `-=`, `*=`, `/=`, etc. | Assignment operators | Non-associative |

```python
1 + (10 * 3) + 3 ** 2
```

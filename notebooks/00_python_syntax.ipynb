## 1. Arthemetic & Operator Precedence 

```python
print(2 + 3 * 6)  # 20 (multiplication evaluated before addition)
print((2 + 3) * 6)  # 30 (parentheses override precedence)
print(2**8)  # 256 (exponentiation)
```


Python Operator Precedence Table

| Precedence  | Operator(s) / Construct                                          | Description                                                             | Associativity |
| ----------- | ---------------------------------------------------------------- | ----------------------------------------------------------------------- | ------------- |
| 1 (Highest) | `()`, `[]`, `{}`                                                 | Grouping, displays, literals                                            | N/A           |
| 2           | `x[index]`, `x[index:index]`, `x(args)`, `x.attribute`           | Subscription, slicing, function calls, attribute references             | Left-to-right |
| 3           | `await x`                                                        | Await expression                                                        | Prefix        |
| 4           | `**`                                                             | Exponentiation                                                          | Right-to-left |
| 5           | `+x`, `-x`, `~x`                                                 | Unary plus, unary minus, bitwise NOT                                    | Prefix        |
| 6           | `*`, `@`, `/`, `//`, `%`                                         | Multiplication, matrix multiplication, division, floor division, modulo | Left-to-right |
| 7           | `+`, `-`                                                         | Addition, subtraction                                                   | Left-to-right |
| 8           | `<<`, `>>`                                                       | Bitwise shifts                                                          | Left-to-right |
| 9           | `&`                                                              | Bitwise AND                                                             | Left-to-right |
| 10          | `^`                                                              | Bitwise XOR                                                             | Left-to-right |
| 11          | `\\\\                                                            | Bitwise OR                                                              | Left-to-right |
| 12          | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `is not`, `in`, `not in` | Comparisons, identity, membership                                       | Chained       |
| 13          | `not x`                                                          | Logical NOT                                                             | Prefix        |
| 14          | `and`                                                            | Logical AND                                                             | Left-to-right |
| 15          | `or`                                                             | Logical OR                                                              | Left-to-right |
| 16          | `x if condition else y`                                          | Conditional expression                                                  | Right-to-left |
| 17          | `lambda`                                                         | Lambda expression                                                       | N/A           |
| 18 (Lowest) | `:=`                                                             | Assignment expression (walrus operator)                                 | N/A           |


```python
# Exponentiation vs unary minus
print(-2**2)       # -4

# Multiplication vs addition
print(2 + 3 * 4)   # 14

# Bitwise AND vs OR
print(5 | 3 & 1)   # 5

# Comparison chaining
print(1 < 2 < 3)   # True

# not vs comparison
print(not 5 == 5)  # False

# and vs or
print(True or False and False)  # True
```

# 6. Comparision, Identity & Boolean Operators

## 7. Conditional Branching: if / elif/ else

- Python uses indentation (4 Spaces per PEP 8) to define block scope.

```python
# Single `if` branch
name: str = "Bob"
if name == "Alice":
    print("Hi, Alice.") # Skipped because condition is False

# `if / else` branching
name: str = "Bob"
if name == "Alice":
    print("Hi, Alice.")
else:
    print("Hello, stranger.")

# `if/elif` branching
name: str = "Bob"
age : int = 30

if name == "Alice":
    print("Hi, Alice")
elif age < 12:
    print("You aren't Alice, kiddo.")
```

```python
# Infinite Loops & Break
# `break` immediately terminates the innermost enclosing loop.
match_name: str = "adam"

def ask_for_name_until_match() -> None:
    """Interactive loop patterns using `while True` and `break`."""

    while True:
        print("Please type your name.")
        name: str = input().lower()
        if name == match_name: 
            break
    print("Hi {}".format(match_name))

ask_for_name_until_match()
```

```python
# Loop termination when condition are no longer true

match_name: str = "smith"
name: str = "default"

while name != match_name:
    name = input().strip().lower()

print("Hey, {}!".format(name))
```

```python
# Loop control with `continue`

def verify_password(match_name: str, match_password: str) -> None:
    """Interactive authentication loop using `continue` and `break`.
    
    Args:
        match_name: (str) Match Name to authorize. 
        match_password: (str) Match Password to authorize.
     
    """
    while True:
        name: str = input().strip()

        if name != match_name:
            continue # Reject non-matching user and restart loop
        print(f"Hello {match_name}, What is the password ?")
        password: str = input().strip()

        if password != match_password:
            print(f"Incorrect Password.\nHint: Password Lenght: {len(match_password)}")
        else:
            break
    print("Acess granted, Welcome back {}.".format(match_name))

user_name: str = "adam"
user_password: str = "swordfish"

verify_password(user_name, user_password)
```

## 11.  Importing Modules & sys.exit()

- Standard `import module` keeps functions namespaced (`random.randint`). 
- `from module import *` imports all public symbols into the local namespace (generally discourged in production code to avoid namespace collisions).
- `sys.exit()`, raises a temporary alert (SystemExit) that lets Python finish cleaning up, like closing files or database connections, before shutting down.
- `sys.exit(0)`: Means the program ran successfully.
- `sys.exit(1)` (or any number other than 0): Means an error occurred.
- `sys.exit("Error message")`: Prints the error message to the screen and automatically sends an error code of 1.
- `raise SystemExit()`, alternative to `sys.exit()`, exactly what `sys.exit()` does, but without needing to `import sys`.


```python

# from random import * # Pollutes local namespace, prefer explict imports.
import random 
import sys
# from typing import Never

for _ in range(5):
    print(random.randint(1, 2)) # Random integers in [1, 10] inclusive

def exit_demo() -> None: # `Never` is more appropriate here.
    """Demonstrates early process termiantion via `sys.exit()`."""
    while True:
        print("Type exit to exit.")
        response: str = input().strip().lower()
        if response == "exit":
            sys.exit(0) # Same as `raise SystemExit(0)`, to terminate the interpreter
        print("You typed: {}".format(response))

exit_demo()
```

Passing and Syntactic Parameter Types

Python provides highly flexible ways to define how arguments map to your parameters when a function is called. 

| Parameter Type | Syntax Example | Description |
| --- | --- | --- |
| **Positional-or-Keyword** | `def func(x):` | The default. Can be passed by order or by name (`func(5)` or `func(x=5)`). |
| **Positional-Only** | `def func(x, /):` | Arguments before the `/` **must** be passed by position only. You cannot use their names. |

| **Arbitrary Keyword (`**kwargs`)** | `def func(**kwargs):` | Dispatches any number of extra keyword arguments into a **dictionary** named `kwargs`. |

## 16. Exception Handling: try / except / finally

- `try` wraps code that may raise an exception.
- `except` catches matching exception types to prevent crashes.
- `finally` executes cleanup code unconditionally, even if `return` or an exception occurs.

```python
def divide_safe(dividend: int, divisor: int) -> float | None:
    try:
        return dividend / divisor
    except ZeroDivisionError as exc:
        print(f"Error: Invalid argument: {exc}")

print(divide_safe(2, 2))  
print(divide_safe(12, 0))  # print(divide_safe(0))  # Prints error message, then returns None implicitly
print(divide_safe(12, 3)) 
```

```python
def divide_safe(dividend: int, divisor: int) -> float | None:
    try:
        return dividend / divisor
    except ZeroDivisionError as exc:
        print(f"Error: Invalid argument: {exc}")
    finally:
        print("--Division finished--") # Runs before the fucntion returns

print(divide_safe(2, 2)) # Prints cleanup, then 1.0 
print(divide_safe(12, 0))  # Prints error, then cleanup, then None
```


## 17. List Indexing & Slicing
- Lists are ordered, mutable sequences. Indexing is 0-based from the left and negative (`-1` is the last element) from the right.
- Slicing `list[start:stop]` returns a new shallow copy from `start` up to `stop` (exclusive).

```python
animals : list[str]= ["cat", "bat", "rat", "elephant"]

print(animals) 
print(animals[0]) 
print(animals[-1]) 

print(animals[0:4]) # ['cat', 'bat', 'rat', 'elephant']
print(animals[1:3]) # ['bat', 'rat']
print(animals[0:-1]) # ['cat', 'bat', 'rat']
```

```python
# Shallow Copying via Full Slice

animals: list[str] = ["cat", "bat", "rat", "elephant"]
animals2 = animals[:]  # Shallow copy (equivalent to `spam.copy()`)

animals3 = animals2 # Just referencing.

print(animals is animals2, id(animals), id(animals2))
print(animals2 is animals3, id(animals2), id(animals3))


animals.append("dog")
print(animals)  # ['cat', 'bat', 'rat', 'elephant', 'dog']
print(animals2)  # ['cat', 'bat', 'rat', 'elephant'] (unaffected)
```

```python
# Length, Index Assignment, Concatenation & Deletion
# Lists support in-place item assignment, `+` / `*` operators, 
# and index removal via `del`.

animals: list[str] = ["cat", "bat", "rat", "elephant"]
animals[1] = "aardvark"
animals[2] = animals[1]
print(animals) # ['cat', 'aardvark', 'aardvark', 'elephant']

print([1, 2, 3] + ["A", "B", "C"]) # [1, 2, 3, 'A', 'B', 'C']
print(["X", "Y", "Z"] * 3) # ['X', 'Y', 'Z', 'X', 'Y', 'Z', 'X', 'Y', 'Z']

animals = ["cat", "bat", "rat", "elephant"]

del animals[2] # Removes element at index 2 ('rat') and shifts subsequent elements
print(animals) # ['cat', 'bat', 'elephant']
```

```python
# List Methods and Sorting


# `.index(x)` returns the first index of `x` (raises ValueError if absent).
# `.count(x)` returns the frequency of `x`.

spam_words: list[str] = ["Zophie", "Pooka", "Fat-tail", "Pooka"]

first_match: int = spam_words.index("Pooka")
print(first_match)  # 1 (first match)

#find the next occurrence, you pass the previous index + 1 as the starting point. 
second_match: int = spam_words.index("Pooka", first_match + 1) 
print(second_match) # 3 (second match)


all_matches: list[int] = [index for index, word in enumerate(spam_words) if word == "Pooka"]

print("All matches: {}".format(all_matches))

print(spam_words.count("Pooka"))  # 2 (counts occurrences)
```

```python

# `.append(x)` adds to the end in O(1) amortized time.
# `.insert(i, x)` inserts at `i` in O(n),if i is invalid index, appends at the end 

spam_words = ["cat", "dog", "bat"]
spam_words.append("moose")  # ['cat', 'dog', 'bat', 'moose']
spam_words.insert(1, "chicken")  # ['cat', 'chicken', 'dog', 'bat', 'moose']
print(spam_words)
```

```python

# `.remove(x)` deletes the first occurrence of `x` (raises ValueError if absent).
# `.pop(i)` removes and returns the item at index `i` (defaults to the last item).

spam_words = ["cat", "bat", "rat", "elephant"]
spam_words.remove("bat")

# Modify existing list (to remove all "bat") Generator + Slice, can use while loop too
spam_words[:] = (word for word in spam_words if word != "bat") 

spam_words = [word for word in spam_words if word!= "bat"] # Create new list

print(spam_words)  # ['cat', 'rat', 'elephant']

popped_item = spam_words.pop(1)  # Removes and returns 'rat'
print(popped_item)  # rat
print(spam_words)  # ['cat', 'elephant']
```

```python

# `.extend(iterable)` appends all elements from an iterable to the end.

print(spam_words)
more_animals = ["monkey", "giraffe"]
spam_words.extend(more_animals)

print(spam_words)  # ['cat', 'elephant', 'monkey', 'giraffe']

# Extend from particular index using slice assignments
spam_words[2:2] = ["tiger", "bear", "camel"] 

print(spam_words)

# New list 
spam_words2 = spam_words[:2] + ["Gorilla", "Rhino"] + spam_words[2:]

print(spam_words2)
```

```python

# `.reverse()` reverses the elements of the list in-place.
# `.reversed()`create new reversed list with list().

spam_words: list[str] = ["cat", "elephant", "monkey", "girafee"] 
spam_words.reverse()
print(spam_words)  # ['giraffe', 'monkey', 'elephant', 'cat']
```

```python

# `.copy()` returns a shallow copy of the list.
copied_list = spam_words.copy()
print(copied_list)  # ['giraffe', 'monkey', 'elephant', 'cat']
```

```python

# `.clear()` empties the list in-place.
copied_list.clear()
print(copied_list)  # [] (The list is now completely empty)
```

```python

# `.sort()` sorts the list in-place and returns `None`
# `sorted()` returns a new sorted list.
numbers: list[int | float] = [2, 5, 3.14, 1, -7]
numbers.sort()
print(numbers)  # [-7, 1, 2, 3.14, 5]

animals: list[str] = ["ants", "cats", "dogs", "badgers", "elephants"]
animals.sort()
print(animals)  # ['ants', 'badgers', 'cats', 'dogs', 'elephants']

animals.sort(reverse=True)
print(animals)  # ['elephants', 'dogs', 'cats', 'badgers', 'ants']

unsorted = ["ants", "cats", "dogs", "badgers", "elephants"]
print(sorted(unsorted))  # Returns new sorted list; `unsorted` remains unchanged
```

## 18. Tuples, Immutable Sequence

- Tuples are ordered, immutable sequences. Because they are immutable,
- tuples containing only hashable elements can be used as dictionary keys or set members.

```python
eggs = ("hello", 42, 0.5)

print(eggs[0])  # 'hello'
print(eggs[1:3])  # (42, 0.5)
print(len(eggs))  # 3

try:
    # Raises TypeError: 'tuple' object does not support item assignment
    eggs[1] = 99 
except TypeError as e:
    print("TypeError:", e)

print(tuple(["cat", "dog", 5]))  # ('cat', 'dog', 5) (list -> tuple conversion)
print(list(("cat", "dog", 5)))  # ['cat', 'dog', 5] (tuple -> list conversion)
```

```python
valid_single_tuple: tuple[str] = ("apple",) # `,` should be included.
invalid_single_tuple = ("orange") # This is just str

print(type(valid_single_tuple))
print(type(invalid_single_tuple))

empty_tuple  = ()
print(type(empty_tuple))

```

```python
# Once a tuple is created, you cannot change its values. Tuples are unchangeable, or immutable as it also is called.
# But there is a workaround. You can convert the tuple into a list, change the list, and convert the list back into a tuple.

x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)

print(x)
```

## 19. Dictionaries (Hash Maps)

- Dictionaries map unique, hashable keys to arbitrary values with O(1) average lookup.
- Insertion order is guaranteed to be preserved.

```python
my_cat: dict[str, str] = {"size": "fat", "color": "gray", "disposition": "loud"}
print(my_cat["size"])  # 'fat'

# Iterating over views: `.values()`, `.keys()`, and `.items()`
cat_dict: dict[str, str | int]  = {"color": "red", "age": 42}
for v in cat_dict.values():
    print(v)  # red, 42

for k in cat_dict.keys():
    print(k)  # color, age

for item in cat_dict.items():
    print(item)  # ('color', 'red'), ('age', 42)

for k, v in cat_dict.items():
    print(f"Key: {k} Value: {v}")
```

```python
# Dictionary Membership, Safe Lookups & Merging 

# `key in dict` checks keys in O(1) time (equivalent to `key in dict.keys()`).
# `{**d1, **d2}` (or `d1 | d2` in Python 3.9+) merges dicts; right-hand keys win on collision.

user_dict: dict[str, str | int] = {"name": "Zophie", "age": 7}
print("name" in user_dict.keys())  # True
print("Zophie" in user_dict.values())  # True (O(n) scan across values)
print("color" in user_dict)  # False (idiomatic O(1) key check)


# `.get(key, default)` avoids `KeyError` by returning `default` (or `None`) when missing.
picnic_items = {"apples": 5, "cups": 2}
print(f"I am bringing {picnic_items.get('cups', 0)} cups.")  # 2
print(f"I am bringing {picnic_items.get('eggs', 0)} eggs.")  # 0 (fallback used)


# `.setdefault(key, default)` inserts `key: default` only if `key` is absent, returning the value.
user_dict = {"name": "Pooka", "age": 5}
print(user_dict.setdefault("color", "black"))  # 'black' (inserts 'color': 'black')
print(
    user_dict.setdefault("color", "white")
)  # 'black' (existing key is not overwritten)

x = {"a": 1, "b": 2}
y = {"b": 3, "c": 4}
z = {**x, **y}  # Or `x | y` in Python 3.9+
print(z)  # {'a': 1, 'b': 3, 'c': 4}

```

```python
# Set algebra: Union (`|`), Intersection (`&`), Difference (`-`), Symmetric Difference (`^`)

set_one, set_two, set_three = {1, 2, 3}, {2, 3, 4}, {3, 4, 5}

print(f"Set One:{set_one}\nSet Two: {set_two}\nSet Three: {set_three}\n")

print("Union of              S1  S2    :", set_one.union(set_two)  ) # elements in either, set_one.union(set_two, set_three) Works too
print("Intersection of       S1  S2  S3:", set_one.intersection(set_two, set_three)) # (element in all)
print("Difference  of        S1  S2    :", set_one.difference(set_two)) # (s1 - s2: in s1 but not s2)
print("Difference of         S2  S1    :", set_two.difference(set_one))  # (s2 - 21: in s2 but not s1))
print("Symmetric Difference  S1  S2    :",set_one.symmetric_difference(set_two)) # (s1 ^ s2: in exactly one set)
```

## 26. Raising Exceptions, Assertions & Logging

- `raise` triggers an exception. Custom exceptions should inherit from `Exception

```python

try:
    raise Exception("This is the error message.")
except Exception as err:
    print(f"Caught: {err}")
```

```python
def box_print(symbol: str, width: int, height: int):
    """Draws an ASCII box after validating input invariants."""

    if len(symbol) != 1:
        raise ValueError("Symbol must be a single character string.")
    if width <= 2:
        raise ValueError("Width must be greater than 2.")
    if height <= 2:
        raise ValueError("Height must be greater than 2.")

    print(symbol * width)
    for _ in range(height - 2):
        print(symbol + (" " * (width - 2)) + symbol)
    print(symbol * width)

try:
    box_print("**", 10, 10)
except ValueError as exc:
    print("ValueError:", exc)
except Exception as exc:
    print("Unexpected Error:",exc)


items = (("*", 4, 4), ("O", 20, 5), ("x", 1, 3), ("ZZ", 3, 3))

for sym, w, h in items:
    try:
        box_print(sym, w, h)
    except Exception as err:
        print(f"An exception happened: {err}")
```

# 28. Structured Logging

- The `logging` module provides severity levels (DEBUG, INFO, WARNING, ERROR, CRITICAL)
- and configurable formatting/handlers without cluttering stdout with ad-hoc `print()` calls.

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format=" %(asctime)s - %(levelname)s - %(message)s",
)
logging.debug("Start of program")


def factorial(n):
    logging.debug("Start of factorial(%s)", n)
    total = 1
    for i in range(1, n + 1):
        total *= i
        logging.debug("i is %s, total is %s", i, total)
    logging.debug("End of factorial(%s)", n)
    return total


number: int = 5
print(factorial(number))  # 120

logging.debug("End of program")
```

## 29. Anonymous Functions (lambda) & Lexical Closures

- `lambda params: expr` creates an inline, single-expression function object.
- A closure occurs when an inner function captures variables from its enclosing scope.

```python
def add(x: int , y: int) -> int:
    """Add the two integers and returns the sum.
   
   Args: 
        x: The first Integar to add.
        y: The second Integar to add.
    
    Returns:
        The sum of x and y.
    
    """

    return x + y


print(add(5, 3))  # 8

print((lambda x, y: x + y)(5, 3))  # 8 (immediately invoked lambda expression)
```

```python
from collections.abc import Callable


def make_adder(n: float) -> Callable[[float], float]:
    """Returns a closure that captures `n` from the enclosing scope."""
    return lambda x: x + float(n)


plus_3:Callable[[float], float] = make_adder(3.0)
plus_5: Callable[[float], float] = make_adder(5)

print(plus_3(4))  # 7
print(plus_5(4))  # 9
```

## 31. Variadic Arguments: *args and **kwargs

- `*args` packs variable positional arguments into a `tuple`.
- `**kwargs` packs variable keyword arguments into a `dict`.


```python
def fruits(*args):
    for fruit in args:
        print(fruit)


fruits("apples", "bananas", "grapes")


def fruit(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")


fruit(name="apple", color="red")


def some_function(*args, **kwargs):
    print(f"Arguments: {args}")
    print(f"Keywords: {kwargs}")


some_function("arg1", "arg2", key1="val1", key2="val2")
```

## 32. Module Entry Point: `if __name__ == "__main__"`

- When a Python file is executed directly, `__name__` is set to `"__main__"`.
- When imported as a module into another file, `__name__` is set to the module's name.
- Guarding execution with this check allows a file to be both importable and runnable.

```python
def add(x: int | float, y: int | float) -> float:
    """Add the input numbers and return the sum as decimal.
    
    Args:
        x: First Number to be added.
        y: Secodn Number to be added.
   
    Returns:
        Returns the sum of x & y as float.  
    """

    return float(x + y)


if __name__ == "__main__":
    print(add(3, 5))  # 8
```

## 1. Arthemetic & Operator Precedence 

```python
print(2 + 3 * 6)  # 20 (multiplication evaluated before addition)
print((2 + 3) * 6)  # 30 (parentheses override precedence)
print(2**8)  # 256 (exponentiation)
```


Python Operator Precedence Table

| Precedence  | Operator(s) / Construct                                          | Description                                                             | Associativity |
| ----------- | ---------------------------------------------------------------- | ----------------------------------------------------------------------- | ------------- |
| 1 (Highest) | `()`, `[]`, `{}`                                                 | Grouping, displays, literals                                            | N/A           |
| 2           | `x[index]`, `x[index:index]`, `x(args)`, `x.attribute`           | Subscription, slicing, function calls, attribute references             | Left-to-right |
| 3           | `await x`                                                        | Await expression                                                        | Prefix        |
| 4           | `**`                                                             | Exponentiation                                                          | Right-to-left |
| 5           | `+x`, `-x`, `~x`                                                 | Unary plus, unary minus, bitwise NOT                                    | Prefix        |
| 6           | `*`, `@`, `/`, `//`, `%`                                         | Multiplication, matrix multiplication, division, floor division, modulo | Left-to-right |
| 7           | `+`, `-`                                                         | Addition, subtraction                                                   | Left-to-right |
| 8           | `<<`, `>>`                                                       | Bitwise shifts                                                          | Left-to-right |
| 9           | `&`                                                              | Bitwise AND                                                             | Left-to-right |
| 10          | `^`                                                              | Bitwise XOR                                                             | Left-to-right |
| 11          | `\|`                                                             | Bitwise OR                                                              | Left-to-right |
| 12          | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `is not`, `in`, `not in` | Comparisons, identity, membership                                       | Chained       |
| 13          | `not x`                                                          | Logical NOT                                                             | Prefix        |
| 14          | `and`                                                            | Logical AND                                                             | Left-to-right |
| 15          | `or`                                                             | Logical OR                                                              | Left-to-right |
| 16          | `x if condition else y`                                          | Conditional expression                                                  | Right-to-left |
| 17          | `lambda`                                                         | Lambda expression                                                       | N/A           |
| 18 (Lowest) | `:=`                                                             | Assignment expression (walrus operator)                                 | N/A           |


```python
# Exponentiation vs unary minus
print(-2**2)       # -4

# Multiplication vs addition
print(2 + 3 * 4)   # 14

# Bitwise AND vs OR
print(5 | 3 & 1)   # 5

# Comparison chaining
print(1 < 2 < 3)   # True

# not vs comparison
print(not 5 == 5)  # False

# and vs or
print(True or False and False)  # True
```

```python
# Loop termination when condition is no longer true

def wait_for_match(match_name: str = "smith") -> None:
    name: str = "default"
    while name != match_name:
        name = input().strip().lower()
    print("Hey, {}!".format(name))


if RUN_INTERACTIVE:
    wait_for_match()
```

```python
# Loop control with `continue`

def verify_password(match_name: str, match_password: str) -> None:
    """Interactive authentication loop using `continue` and `break`.
    
    Args:
        match_name: (str) Match Name to authorize. 
        match_password: (str) Match Password to authorize.
     
    """
    while True:
        name: str = input().strip()

        if name != match_name:
            continue # Reject non-matching user and restart loop
        print(f"Hello {match_name}, What is the password ?")
        password: str = input().strip()

        if password != match_password:
            print(f"Incorrect Password.\nHint: Password Lenght: {len(match_password)}")
        else:
            break
    print("Acess granted, Welcome back {}.".format(match_name))

user_name: str = "adam"
user_password: str = "swordfish"

if RUN_INTERACTIVE:
    verify_password(user_name, user_password)
```

## 11.  Importing Modules & sys.exit()

- Standard `import module` keeps functions namespaced (`random.randint`). 
- `from module import *` imports all public symbols into the local namespace (generally discourged in production code to avoid namespace collisions).
- `sys.exit()`, raises a temporary alert (SystemExit) that lets Python finish cleaning up, like closing files or database connections, before shutting down.
- `sys.exit(0)`: Means the program ran successfully.
- `sys.exit(1)` (or any number other than 0): Means an error occurred.
- `sys.exit("Error message")`: Prints the error message to the screen and automatically sends an error code of 1.
- `raise SystemExit()`, alternative to `sys.exit()`, exactly what `sys.exit()` does, but without needing to `import sys`.


```python

# from random import * # Pollutes local namespace, prefer explict imports.
import random 
import sys
# from typing import Never

for _ in range(5):
    print(random.randint(1, 2)) # Random integers in [1, 2] inclusive

def exit_demo() -> None: # `Never` is more appropriate here.
    """Demonstrates early process termiantion via `sys.exit()`."""
    while True:
        print("Type exit to exit.")
        response: str = input().strip().lower()
        if response == "exit":
            sys.exit(0) # Same as `raise SystemExit(0)`, to terminate the interpreter
        print("You typed: {}".format(response))

if RUN_INTERACTIVE:
    exit_demo()
```

| Parameter Type | Syntax Example | Description |

## 16. Exception Handling: try / except / finally

- `try` wraps code that may raise an exception.
- `except` catches matching exception types to prevent crashes.
- `finally` executes cleanup code unconditionally, even if `return` or an exception occurs.

```python
def divide_safe(dividend: int, divisor: int) -> float | None:
    try:
        return dividend / divisor
    except ZeroDivisionError as exc:
        print(f"Error: Invalid argument: {exc}")

print(divide_safe(2, 2))  
print(divide_safe(12, 0))  # print(divide_safe(0))  # Prints error message, then returns None implicitly
print(divide_safe(12, 3)) 
```

```python
def divide_safe(dividend: int, divisor: int) -> float | None:
    try:
        return dividend / divisor
    except ZeroDivisionError as exc:
        print(f"Error: Invalid argument: {exc}")
    finally:
        print("--Division finished--") # Runs before the fucntion returns

print(divide_safe(2, 2)) # Prints cleanup, then 1.0 
print(divide_safe(12, 0))  # Prints error, then cleanup, then None
```


## 17. List Indexing & Slicing
- Lists are ordered, mutable sequences. Indexing is 0-based from the left and negative (`-1` is the last element) from the right.
- Slicing `list[start:stop]` returns a new shallow copy from `start` up to `stop` (exclusive).

```python
animals : list[str]= ["cat", "bat", "rat", "elephant"]

print(animals) 
print(animals[0]) 
print(animals[-1]) 

print(animals[0:4]) # ['cat', 'bat', 'rat', 'elephant']
print(animals[1:3]) # ['bat', 'rat']
print(animals[0:-1]) # ['cat', 'bat', 'rat']
```

```python
# Shallow Copying via Full Slice

animals: list[str] = ["cat", "bat", "rat", "elephant"]
animals2 = animals[:]  # Shallow copy (equivalent to `spam.copy()`)

animals3 = animals2 # Just referencing.

print(animals is animals2, id(animals), id(animals2))
print(animals2 is animals3, id(animals2), id(animals3))


animals.append("dog")
print(animals)  # ['cat', 'bat', 'rat', 'elephant', 'dog']
print(animals2)  # ['cat', 'bat', 'rat', 'elephant'] (unaffected)
```

```python
# Length, Index Assignment, Concatenation & Deletion
# Lists support in-place item assignment, `+` / `*` operators, 
# and index removal via `del`.

animals: list[str] = ["cat", "bat", "rat", "elephant"]
animals[1] = "aardvark"
animals[2] = animals[1]
print(animals) # ['cat', 'aardvark', 'aardvark', 'elephant']

print([1, 2, 3] + ["A", "B", "C"]) # [1, 2, 3, 'A', 'B', 'C']
print(["X", "Y", "Z"] * 3) # ['X', 'Y', 'Z', 'X', 'Y', 'Z', 'X', 'Y', 'Z']

animals = ["cat", "bat", "rat", "elephant"]

del animals[2] # Removes element at index 2 ('rat') and shifts subsequent elements
print(animals) # ['cat', 'bat', 'elephant']
```

```python
# Iteration Helpers: enumerate() & zip() 

# `enumerate(iterable)` yields `(index, item)` tuples.
supplies = ["pens", "staplers", "flame-throwers", "binders"]
for i, supply in enumerate(supplies):
    print(f"Index {i} in supplies is: {supply}")
```

```python

# `zip(*iterables)` yields tuples pairing elements at matching indices across iterables,
# stopping when the shortest iterable is exhausted.
names: list[str] = ["Pete", "John", "Elizabeth"]
ages: list[int] = [6, 23, 44]
for n, a in zip(names, ages):
    print(f"{n} is {a} years old")
```

```python
# Membership Testing,`in` / `not in` perform linear O(n) membership scans on lists

words: list[str] = ["hello", "hi", "howdy", "hayas"]

print("howdy" in words)
print("cat" in words)
print("howdy" not in words)
```

```python
# List Methods and Sorting


# `.index(x)` returns the first index of `x` (raises ValueError if absent).
# `.count(x)` returns the frequency of `x`.

spam_words: list[str] = ["Zophie", "Pooka", "Fat-tail", "Pooka"]

first_match: int = spam_words.index("Pooka")
print(first_match)  # 1 (first match)

#find the next occurrence, you pass the previous index + 1 as the starting point. 
second_match: int = spam_words.index("Pooka", first_match + 1) 
print(second_match) # 3 (second match)


all_matches: list[int] = [index for index, word in enumerate(spam_words) if word == "Pooka"]

print("All matches: {}".format(all_matches))

print(spam_words.count("Pooka"))  # 2 (counts occurrences)
```

```python

# `.append(x)` adds to the end in O(1) amortized time.
# `.insert(i, x)` inserts at `i` in O(n),if i is invalid index, appends at the end 

spam_words = ["cat", "dog", "bat"]
spam_words.append("moose")  # ['cat', 'dog', 'bat', 'moose']
spam_words.insert(1, "chicken")  # ['cat', 'chicken', 'dog', 'bat', 'moose']
print(spam_words)
```

```python

# `.remove(x)` deletes the first occurrence of `x` (raises ValueError if absent).
# `.pop(i)` removes and returns the item at index `i` (defaults to the last item).

spam_words = ["cat", "bat", "rat", "elephant"]
spam_words.remove("bat")

# Modify existing list (to remove all "bat") Generator + Slice, can use while loop too
spam_words[:] = (word for word in spam_words if word != "bat") 

spam_words = [word for word in spam_words if word!= "bat"] # Create new list

print(spam_words)  # ['cat', 'rat', 'elephant']

popped_item = spam_words.pop(1)  # Removes and returns 'rat'
print(popped_item)  # rat
print(spam_words)  # ['cat', 'elephant']
```

```python

# `.extend(iterable)` appends all elements from an iterable to the end.

print(spam_words)
more_animals = ["monkey", "giraffe"]
spam_words.extend(more_animals)

print(spam_words)  # ['cat', 'elephant', 'monkey', 'giraffe']

# Extend from particular index using slice assignments
spam_words[2:2] = ["tiger", "bear", "camel"] 

print(spam_words)

# New list 
spam_words2 = spam_words[:2] + ["Gorilla", "Rhino"] + spam_words[2:]

print(spam_words2)
```

```python

# `.reverse()` reverses the elements of the list in-place.
# `reversed(seq)` is a built-in returning a reverse iterator; wrap it in list() for a new list

spam_words: list[str] = ["cat", "elephant", "monkey", "giraffe"] 
spam_words.reverse()
print(spam_words)  # ['giraffe', 'monkey', 'elephant', 'cat']
print(list(reversed(spam_words)))  # ['cat', 'elephant', 'monkey', 'giraffe']
```

```python

# `.copy()` returns a shallow copy of the list.
copied_list = spam_words.copy()
print(copied_list)  # ['giraffe', 'monkey', 'elephant', 'cat']
```

```python

# `.clear()` empties the list in-place.
copied_list.clear()
print(copied_list)  # [] (The list is now completely empty)
```

```python

# `.sort()` sorts the list in-place and returns `None`
# `sorted()` returns a new sorted list.
numbers: list[int | float] = [2, 5, 3.14, 1, -7]
numbers.sort()
print(numbers)  # [-7, 1, 2, 3.14, 5]

animals: list[str] = ["ants", "cats", "dogs", "badgers", "elephants"]
animals.sort()
print(animals)  # ['ants', 'badgers', 'cats', 'dogs', 'elephants']

animals.sort(reverse=True)
print(animals)  # ['elephants', 'dogs', 'cats', 'badgers', 'ants']

unsorted = ["ants", "cats", "dogs", "badgers", "elephants"]
print(sorted(unsorted))  # Returns new sorted list; `unsorted` remains unchanged
```

## 18. Tuples, Immutable Sequence

- Tuples are ordered, immutable sequences. Because they are immutable,
- tuples containing only hashable elements can be used as dictionary keys or set members.

```python
eggs = ("hello", 42, 0.5)

print(eggs[0])  # 'hello'
print(eggs[1:3])  # (42, 0.5)
print(len(eggs))  # 3

try:
    # Raises TypeError: 'tuple' object does not support item assignment
    eggs[1] = 99 
except TypeError as e:
    print("TypeError:", e)

print(tuple(["cat", "dog", 5]))  # ('cat', 'dog', 5) (list -> tuple conversion)
print(list(("cat", "dog", 5)))  # ['cat', 'dog', 5] (tuple -> list conversion)
```

```python
valid_single_tuple: tuple[str] = ("apple",) # `,` should be included.
invalid_single_tuple = ("orange") # This is just str

print(type(valid_single_tuple))
print(type(invalid_single_tuple))

empty_tuple  = ()
print(type(empty_tuple))

```

```python
# Once a tuple is created, you cannot change its values. Tuples are unchangeable, or immutable as it also is called.
# But there is a workaround. You can convert the tuple into a list, change the list, and convert the list back into a tuple.

x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)

print(x)
```

## 19. Dictionaries (Hash Maps)

- Dictionaries map unique, hashable keys to arbitrary values with O(1) average lookup.
- Insertion order is guaranteed to be preserved.

```python
my_cat: dict[str, str] = {"size": "fat", "color": "gray", "disposition": "loud"}
print(my_cat["size"])  # 'fat'

# Iterating over views: `.values()`, `.keys()`, and `.items()`
cat_dict: dict[str, str | int]  = {"color": "red", "age": 42}
for v in cat_dict.values():
    print(v)  # red, 42

for k in cat_dict.keys():
    print(k)  # color, age

for item in cat_dict.items():
    print(item)  # ('color', 'red'), ('age', 42)

for k, v in cat_dict.items():
    print(f"Key: {k} Value: {v}")
```

```python
# Dictionary Membership, Safe Lookups & Merging 

# `key in dict` checks keys in O(1) time (equivalent to `key in dict.keys()`).
# `{**d1, **d2}` (or `d1 | d2` in Python 3.9+) merges dicts; right-hand keys win on collision.

user_dict: dict[str, str | int] = {"name": "Zophie", "age": 7}
print("name" in user_dict.keys())  # True
print("Zophie" in user_dict.values())  # True (O(n) scan across values)
print("color" in user_dict)  # False (idiomatic O(1) key check)


# `.get(key, default)` avoids `KeyError` by returning `default` (or `None`) when missing.
picnic_items = {"apples": 5, "cups": 2}
print(f"I am bringing {picnic_items.get('cups', 0)} cups.")  # 2
print(f"I am bringing {picnic_items.get('eggs', 0)} eggs.")  # 0 (fallback used)


# `.setdefault(key, default)` inserts `key: default` only if `key` is absent, returning the value.
user_dict = {"name": "Pooka", "age": 5}
print(user_dict.setdefault("color", "black"))  # 'black' (inserts 'color': 'black')
print(
    user_dict.setdefault("color", "white")
)  # 'black' (existing key is not overwritten)

x = {"a": 1, "b": 2}
y = {"b": 3, "c": 4}
z = {**x, **y}  # Or `x | y` in Python 3.9+
print(z)  # {'a': 1, 'b': 3, 'c': 4}

```

## 26. Raising Exceptions, Assertions & Logging

```python
def box_print(symbol: str, width: int, height: int):
    """Draws an ASCII box after validating input invariants."""

    if len(symbol) != 1:
        raise ValueError("Symbol must be a single character string.")
    if width <= 2:
        raise ValueError("Width must be greater than 2.")
    if height <= 2:
        raise ValueError("Height must be greater than 2.")

    print(symbol * width)
    for _ in range(height - 2):
        print(symbol + (" " * (width - 2)) + symbol)
    print(symbol * width)

try:
    box_print("**", 10, 10)
except ValueError as exc:
    print("ValueError:", exc)
except Exception as exc:
    print("Unexpected Error:",exc)


items = (("*", 4, 4), ("O", 20, 5), ("x", 1, 3), ("ZZ", 3, 3))

for sym, w, h in items:
    try:
        box_print(sym, w, h)
    except Exception as err:
        print(f"An exception happened: {err}")
```

## 38. Mutability, Aliasing & Deep Copy

- Immutable: `int, float, str, tuple, frozenset, bytes`. Mutable: `list, dict, set, bytearray`.
- Assignment never copies. `.copy()` / `[:]` are shallow; `copy.deepcopy` recurses.

```python
import copy
 
matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
deep = copy.deepcopy(matrix)
matrix[0][0] = 99
print(shallow[0][0], deep[0][0])  # 99 1 (shallow shares inner lists)
 
# Classic gotcha: repetition copies references, not objects.
bad = [[0] * 3] * 3
bad[0][0] = 1
print(bad)  # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]
good = [[0] * 3 for _ in range(3)]
good[0][0] = 1
print(good)  # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
 
# Tuples are immutable, but a mutable item inside can still change.
t = ([1], 2)
t[0].append(5)
print(t)  # ([1, 5], 2)
```

## 42. Scope: LEGB, `nonlocal` & Closure Late Binding

- Lookup order: **L**ocal -> **E**nclosing -> **G**lobal -> **B**uilt-in.


```python
def counter() -> "callable":
    count = 0
 
    def inc() -> int:
        nonlocal count  # rebind the enclosing variable (global would be wrong here)
        count += 1
        return count
 
    return inc
 
c = counter()
print(c(), c(), c())  # 1 2 3
 
# Late binding: closures capture the variable, not its value at creation time.
funcs = [lambda: i for i in range(3)]
print([f() for f in funcs])  # [2, 2, 2]
funcs = [lambda i=i: i for i in range(3)]  # default arg freezes the value
print([f() for f in funcs])  # [0, 1, 2]
```

## 43. Decorators
- A decorator is a callable that takes a function and returns a replacement.
- `@deco` above `def f` is sugar for `f = deco(f)`.

```python
import functools
import time
from collections.abc import Callable
from typing import Any
 
 
def timer(fn: Callable[..., Any]) -> Callable[..., Any]:
    @functools.wraps(fn)  # preserves __name__, __doc__
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        start = time.perf_counter()
        try:
            return fn(*args, **kwargs)
        finally:
            print(f"{fn.__name__} took {time.perf_counter() - start:.6f}s")
 
    return wrapper
 
 
def repeat(times: int) -> Callable[[Callable[..., Any]], Callable[..., Any]]:
    """Decorator factory: three nested levels."""
 
    def decorator(fn: Callable[..., Any]) -> Callable[..., Any]:
        @functools.wraps(fn)
        def wrapper(*args: Any, **kwargs: Any) -> Any:
            result = None
            for _ in range(times):
                result = fn(*args, **kwargs)
            return result
 
        return wrapper
 
    return decorator
 
 
@timer
def slow_sum(n: int) -> int:
    """Sum 0..n-1."""
    return sum(range(n))
 
 
@repeat(times=2)
@timer  # stacked: applied bottom-up
def hello(name: str) -> str:
    print(f"Hello {name}")
    return name
 
 
print(slow_sum(100_000), slow_sum.__name__)
hello("Ada")
```

## 45. Iterators & the Iteration Protocol
- Iterable: has `__iter__`. Iterator: has `__iter__` + `__next__`; raises `StopIteration`.
- `for` calls `iter()` once, then `next()` repeatedly.


```python
it = iter([10, 20])
print(next(it), next(it))
print(next(it, "done"))  # default instead of StopIteration
try:
    next(it)
except StopIteration:
    print("exhausted")
 
 
class Countdown:
    """A hand-written iterator."""
 
    def __init__(self, start: int) -> None:
        self.current = start
 
    def __iter__(self) -> "Countdown":
        return self
 
    def __next__(self) -> int:
        if self.current <= 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1
 
 
print(list(Countdown(3)))  # [3, 2, 1]
 
# Iterators are single-use; sequences can be iterated repeatedly.
gen = (x for x in range(3))
print(list(gen), list(gen))  # [0, 1, 2] []
 
# Two-argument iter(callable, sentinel): call until sentinel is returned.
import io
stream = io.StringIO("a\nb\n\n")
print(list(iter(stream.readline, "\n")))  # stops at first blank line -> ['a\n', 'b\n']
```

## 46. Generators: `yield`, `yield from`, `send`
- Any function containing `yield` returns a lazy generator object.

```python
from collections.abc import Generator, Iterator 
from itertools import islice
 
def evens(limit: int) -> Iterator[int]:
    for n in range(0, limit, 2):
        yield n  # execution pauses here and resumes on next()
 
print(list(evens(10)))
 
 
def flatten(nested: list) -> Iterator:
    for item in nested:
        if isinstance(item, list):
            yield from flatten(item)  # delegate to a sub-generator
        else:
            yield item
 
print(list(flatten([1, [2, [3, 4]], 5])))
 
 
def accumulator() -> Generator[int, int, str]:
    total = 0
    while True:
        value = yield total  # receives values via .send()
        if value is None:
            return "finished"
        total += value
 
acc = accumulator()
next(acc)  # prime the generator
print(acc.send(5), acc.send(10))  # 5 15
 
# Generator expressions avoid building the whole list.
print(sum(x * x for x in range(1_000_000)))
 
 
def fib_gen() -> Iterator[int]:
    a, b = 0, 1
    while True:  # infinite generators are fine when consumed lazily
        yield a
        a, b = b, a + b
 
print(list(islice(fib_gen(), 10)))
```

## 53. Exceptions: `else`, Multiple Types, Chaining, Groups

```python

def parse(text: str) -> int:
    try:
        value = int(text)
    except (ValueError, TypeError) as exc:  # several types in one clause
        print(f"bad input {text!r}: {exc}")
        return -1
    except Exception:
        raise  # bare raise re-raises the active exception unchanged
    else:  # runs ONLY if no exception occurred
        print("parsed fine")
        return value
    finally:  # always runs
        print("cleanup")
 
print(parse("42"), parse("x"), parse(None))  # type: ignore[arg-type]
```

## 54. Context Managers & `with`
- `with` guarantees cleanup via `__enter__` / `__exit__`, even on exceptions.

```python

from contextlib import contextmanager, suppress, ExitStack, redirect_stdout
 
 
class Timer:
    def __enter__(self) -> "Timer":
        self.start = time.perf_counter()
        return self  # bound by `as`
 
    def __exit__(self, exc_type, exc, tb) -> bool:
        print(f"elapsed: {time.perf_counter() - self.start:.4f}s")
        return False  # False/None -> propagate exceptions; True -> suppress
 
with Timer() as t:
    sum(range(10_000))
 
 
@contextmanager
def tag(name: str):
    print(f"<{name}>")
    try:
        yield name  # value bound by `as`
    finally:
        print(f"</{name}>")
 
with tag("html") as tag_name:
    print("inside", tag_name)
 
with suppress(FileNotFoundError):  # replaces try/except/pass
    open("does_not_exist.txt")
 
buffer = io.StringIO()
with redirect_stdout(buffer):
    print("captured")
print(repr(buffer.getvalue()))
 
# Several managers at once (parenthesized form is 3.10+)
with (
    tag("a"),
    tag("b"),
):
    print("nested")
 
```

##  56. Classes: Attributes, Methods, Properties
- `self` is the instance. Class attributes are shared; instance attributes are per-object.

```python
class Account:
    bank = "PyBank"  # class attribute (shared)
    _count = 0
 
    def __init__(self, owner: str, balance: float = 0.0) -> None:
        self.owner = owner  # instance attributes
        self._balance = balance  # single `_`: "internal" by convention
        self.__secret = 42  # double `__`: name-mangled to _Account__secret
        Account._count += 1
 
    @property
    def balance(self) -> float:  # read access looks like an attribute
        return self._balance
 
    @balance.setter
    def balance(self, value: float) -> None:
        if value < 0:
            raise ValueError("balance cannot be negative")
        self._balance = value
 
    def deposit(self, amount: float) -> "Account":
        self.balance += amount
        return self  # returning self enables chaining
 
    @classmethod
    def from_string(cls, text: str) -> "Account":  # alternative constructor
        owner, bal = text.split(":")
        return cls(owner, float(bal))
 
    @staticmethod
    def validate(amount: float) -> bool:  # no self/cls; a namespaced function
        return amount > 0
 
 
acct = Account.from_string("Ada:100").deposit(50).deposit(25)
print(acct.owner, acct.balance, Account.bank, Account._count, Account.validate(5))
print(acct._Account__secret)  # mangled name (not real privacy)
try:
    acct.balance = -1
except ValueError as exc:
    print("ValueError:", exc)
 
acct.extra = "dynamic attributes are allowed by default"
print(acct.__dict__.keys())
```

## 57. Inheritance, `super()`, MRO, Mixins

```python

class Animal:
    def __init__(self, name: str) -> None:
        self.name = name
 
    def speak(self) -> str:
        return "..."
 
    def __repr__(self) -> str:
        return f"{type(self).__name__}({self.name!r})"
 
 
class Dog(Animal):
    def __init__(self, name: str, breed: str) -> None:
        super().__init__(name)  # call the parent initializer
        self.breed = breed
 
    def speak(self) -> str:  # override
        return "Woof"
 
 
class LoudMixin:
    def speak(self) -> str:
        return super().speak().upper() + "!"  # cooperates in the MRO chain
 
 
class LoudDog(LoudMixin, Dog):
    pass
 
 
ld = LoudDog("Rex", "lab")
print(ld, ld.speak())  # LoudDog('Rex') WOOF!
print([c.__name__ for c in LoudDog.__mro__])  # method resolution order
print(isinstance(ld, Animal), issubclass(LoudDog, Dog))
```

## 58. Dunder (Magic) Methods

```python

from functools import total_ordering
 
 
@total_ordering  # derive <=, >, >= from __eq__ and __lt__
class Version:
    def __init__(self, *parts: int) -> None:
        self.parts = tuple(parts)
 
    def __repr__(self) -> str:  # unambiguous, for developers
        return f"Version{self.parts}"
 
    def __str__(self) -> str:  # readable, for users
        return ".".join(map(str, self.parts))
 
    def __eq__(self, other: object) -> bool:
        return isinstance(other, Version) and self.parts == other.parts
 
    def __lt__(self, other: "Version") -> bool:
        return self.parts < other.parts
 
    def __hash__(self) -> int:  # define whenever __eq__ is defined
        return hash(self.parts)
 
    def __len__(self) -> int:
        return len(self.parts)
 
    def __getitem__(self, i):  # indexing + slicing + iteration fallback
        return self.parts[i]
 
    def __contains__(self, item: int) -> bool:
        return item in self.parts
 
    def __add__(self, other: "Version") -> "Version":  # operator overloading
        return Version(*(a + b for a, b in zip(self.parts, other.parts)))
 
    def __bool__(self) -> bool:
        return any(self.parts)
 
    def __call__(self, sep: str = "-") -> str:  # instances become callable
        return sep.join(map(str, self.parts))


v1, v2 = Version(1, 2, 3), Version(1, 10, 0)
print(str(v1), repr(v1), v1 < v2, v1 >= v2, v1 == Version(1, 2, 3))
print(len(v1), v1[0], v1[1:], 2 in v1, list(v1), v1 + v2, v1("_"))


print(sorted([v2, v1]), {v1: "hashable"}, bool(Version(0, 0)))
print(max(v1, v2))


# Context-manager, iterator and descriptor protocols are also just dunders:
# __enter__/__exit__, __iter__/__next__, __get__/__set__, __getattr__, __setattr__,
# __class_getitem__, __init_subclass__, __post_init__ (dataclasses).
```

## 59. `dataclass`, `Enum`, `ABC`, `__slots__`

```python

from abc import ABC, abstractmethod
from dataclasses import dataclass, field, asdict, replace
from enum import Enum, IntEnum, auto, Flag
 
 
@dataclass(order=True, frozen=True)  # generates __init__, __repr__, __eq__, ordering, hash
class Item:
    name: str
    price: float = 0.0
    tags: tuple[str, ...] = ()
 
 
@dataclass
class Cart:
    items: list[Item] = field(default_factory=list)  # never use a bare mutable default
    total: float = field(init=False, default=0.0)
 
    def __post_init__(self) -> None:
        self.total = sum(i.price for i in self.items)
 
 
cart = Cart([Item("pen", 1.5), Item("book", 12.0, ("paper",))])
print(cart)
print(asdict(cart.items[1]), replace(cart.items[0], price=2.0))
try:
    cart.items[0].price = 9  # type: ignore[misc]
except Exception as exc:
    print(type(exc).__name__)  # FrozenInstanceError
```

```python
 
class Color(Enum):
    RED = 1
    GREEN = auto()  # auto-assigned: 2
 
class Level(IntEnum):  # also behaves like int
    LOW = 1
    HIGH = 2
 
class Perm(Flag):
    R = auto()
    W = auto()
 
print(Color.RED, Color.RED.name, Color.RED.value, Color(2), Color["GREEN"])
print(list(Color), Level.HIGH > Level.LOW, Perm.R | Perm.W)
 
 
class Shape(ABC):
    @abstractmethod
    def area(self) -> float: ...  # `...` is a valid placeholder body
 
class Square(Shape):
    __slots__ = ("side",)  # no per-instance __dict__; saves memory, blocks new attrs
 
    def __init__(self, side: float) -> None:
        self.side = side
 
    def area(self) -> float:
        return self.side**2
 
try:
    Shape()  # type: ignore[abstract]
except TypeError as exc:
    print("TypeError:", exc)
print(Square(3).area()) 
```

## 61. Structural Pattern Matching: All Pattern Kinds

```python

@dataclass
class Click:
    x: int
    y: int
 
def handle(event: Any) -> str:
    match event:
        case 0 | 1 | 2:  # literals with OR
            return "small int"
        case "quit" | "exit":
            return "quit"
        case str() as text if text.startswith("#"):  # class pattern + `as` + guard
            return f"comment {text}"
        case [x]:
            return f"one item {x}"
        case [first, *rest]:  # sequence with star capture
            return f"first={first}, rest={rest}"
        case {"type": "key", "key": k, **others}:  # mapping; extra keys via **
            return f"key {k} {others}"
        case Click(x=0, y=0):  # class pattern with attribute patterns
            return "origin click"
        case Click(x, y):  # positional (dataclasses provide __match_args__)
            return f"click at {x},{y}"
        case Color.RED:  # dotted name = value pattern
            return "red"
        case None:
            return "nothing"
        case int(n) | float(n):  # type pattern, both binding the same name
            return f"number {n}"
        case _:  # wildcard
            return "unknown"
 
for ev in [1, "exit", "#note", [7], [1, 2, 3], {"type": "key", "key": "a", "mod": 1},
           Click(0, 0), Click(3, 4), Color.RED, None, 9.5, object()]:
    print(f"{ev!r:.40} -> {handle(ev)}")
```

## 63. `async` / `await` with `asyncio`
- Coroutines (`async def`) run on an event loop and cooperatively yield at `await`.
- Great for many concurrent I/O waits; not for CPU-bound work.

```python
import asyncio
 
 
async def fetch(name: str, delay: float) -> str:
    await asyncio.sleep(delay)  # non-blocking pause
    return f"{name} done"
 
 
async def ticker(n: int):
    for i in range(n):
        await asyncio.sleep(0)
        yield i  # async generator
 
 
class AsyncCM:
    async def __aenter__(self) -> str:
        return "entered"
 
    async def __aexit__(self, *exc) -> None:
        print("async exit")
 
 
async def main() -> None:
    results = await asyncio.gather(fetch("a", 0.02), fetch("b", 0.01))  # run concurrently
    print(results)
 
    async for i in ticker(3):
        print("tick", i)
 
    async with AsyncCM() as v:
        print(v)
 
    print([i async for i in ticker(3)])  # async comprehension
 
    try:
        await asyncio.wait_for(fetch("slow", 1), timeout=0.01)
    except asyncio.TimeoutError:
        print("timed out")
 
    async with asyncio.TaskGroup() as tg:  # structured concurrency (3.11+)
        t1 = tg.create_task(fetch("x", 0.01))
        t2 = tg.create_task(fetch("y", 0.01))
    print(t1.result(), t2.result())
 
 
# `asyncio.run` fails inside an already running loop (Jupyter). Use `await main()` there.
try:
    asyncio.get_running_loop()
    print("loop already running: use `await main()` in this environment")
except RuntimeError:
    asyncio.run(main())
```

## 64. Regular Expressions (`re`)

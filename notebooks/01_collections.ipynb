## Python Collections (Arrays)

There are four collection data types in the Python programming language:

-   **[List](https://www.w3schools.com/python/python_lists.asp)** is a collection which is ordered and changeable. Allows duplicate members.
-   **[Tuple](https://www.w3schools.com/python/python_tuples.asp)** is a collection which is ordered and unchangeable. Allows duplicate members.
-   **[Set](https://www.w3schools.com/python/python_sets.asp)** is a collection which is unordered, unchangeable\*, and unindexed. No duplicate members.
-   **[Dictionary](https://www.w3schools.com/python/python_dictionaries.asp)** is a collection which is ordered\*\* and changeable. No duplicate members.

## **Quick Comparison**

| Property | List | Tuple | Set | Dictionary | Frozenset |
| --- | --- | --- | --- | --- | --- |
| Syntax | `[]` | `()` | `{}` | `{key: value}` | `frozenset()` |
| Ordered | Yes | Yes | No  | Yes* | No  |
| Mutable | Yes | No  | Yes | Yes | No  |
| Allows duplicates | Yes | Yes | No  | Keys: No, Values: Yes | No  |
| Indexing | Yes | Yes | No  | By key | No  |
| Slicing | Yes | Yes | No  | No  | No  |
| Mixed data types | Yes | Yes | Yes** | Yes | Yes** |
| Nested collections | Yes | Yes | Hashable only | Yes | Hashable only |
| Hashable | No  | Sometimes | No  | No  | Yes |

* Dictionary insertion order is guaranteed in Python 3.7 and later. **Set and frozenset elements must be hashable.


-   **Ordered ≠ Immutable:** Lists are ordered but mutable; tuples are ordered and immutable.
-   **Unordered ≠ Unchangeable:** Sets are unordered but mutable.
-   **Immutable ≠ Always Hashable:** A tuple containing a list is not hashable.
-   **Dictionary keys must be unique and hashable**, but dictionary values can be duplicated and can have any data type.
-   **Sets and dictionaries use hashing** for efficient membership and lookup operations.
-   **Empty collection syntax matters:**
    -   Empty list: `[]`
    -   Empty tuple: `()`
    -   Empty set: `set()` (not `{}`)
    -   Empty dictionary: `{}`

```python
# Imports

from typing import Any
```

```python
# Helper Functions

def print_list(any_list: list[Any]) -> None:
    """Print input list with indicies.

    Args:
        any_list: (Any) List to be printed with indicies.
    
    """
    print(*(f"{index}: {element}\n" for index, element in enumerate(any_list)), sep="")
```

## **1\. List (**`list`**)**

Example: `my_list = [10, 20, 30, 20]`

-   **Ordered:** Maintains the order in which elements are inserted.
    
-   **Mutable:** Elements can be added, modified, or removed.
    
-   **Allows duplicates:** The same value can appear multiple times.
    
-   **Allows mixed data types:** Can contain integers, strings, floats, booleans, and other objects.
    
-   **Indexable:** Elements can be accessed using their index.
    
-   **Supports negative indexing:** `-1` accesses the last element.
    
-   **Supports slicing:** Can extract a portion of the list.
    
-   **Iterable:** Can be traversed using loops.
    
-   **Allows nested collections:** A list can contain other lists, tuples, dictionaries, etc.
    
-   **Dynamic size:** Automatically grows or shrinks as elements are added or removed.
    
-   **Not hashable:** Cannot be used as a dictionary key or set element.

## **2\. Tuple (**`tuple`**)**

Example: `my_tuple = (10, 20, 30, 20)`

-   **Ordered:** Maintains the order in which elements are inserted.
    
-   **Immutable:** Elements cannot be added, modified, or removed after creation.
    
-   **Allows duplicates:** The same value can appear multiple times.
    
-   **Allows mixed data types:** Can contain different types of objects.
    
-   **Indexable:** Elements can be accessed using their index.
    
-   **Supports negative indexing:** `-1` accesses the last element.
    
-   **Supports slicing:** Can extract a portion of the tuple.
    
-   **Iterable:** Can be traversed using loops.
    
-   **Allows nested collections:** Can contain lists, dictionaries, other tuples, etc.
    
-   **Fixed size:** Cannot change its number of elements after creation.
    
-   **Hashable when all elements are hashable:** Such tuples can be used as dictionary keys or set elements.
    
-   **Generally uses less memory than a list** for comparable contents.

## **3\. Set (**`set`**)**

Example: `my_set = {10, 20, 30, 20}`

-   **Unordered:** Does not maintain a guaranteed insertion order.
    
-   **Mutable:** Elements can be added or removed.
    
-   **Does not allow duplicates:** Duplicate values are automatically removed.
    
-   **Allows mixed data types:** Can contain different hashable object types.
    
-   **Not indexable:** Elements cannot be accessed using indexes.
    
-   **Does not support slicing:** Cannot extract elements using slicing.
    
-   **Iterable:** Can be traversed using loops.
    
-   **No nested mutable collections:** Elements must be hashable, so lists and dictionaries cannot be stored directly.
    
-   **Dynamic size:** Can grow or shrink.
    
-   **Stores only unique hashable elements.**
    
-   **Supports mathematical set operations:**
    
    -   Union (`|`): Combines elements from both sets.
        
    -   Intersection (`&`): Returns common elements.
        
    -   Difference (`-`): Returns elements present in the first set but not the second.
        
    -   Symmetric difference (`^`): Returns elements present in either set, but not both.
        
    
-   **Membership testing:** Checking whether an element exists is generally very fast.

## **4\. Dictionary (**`dict`**)**

Example: `my_dict = {"name": "John", "age": 25}`

-   **Insertion ordered:** Maintains the order in which keys are inserted (guaranteed in Python 3.7+).
    
-   **Mutable:** Keys can be added, values modified, and entries removed.
    
-   **Stores key-value pairs:** Each entry contains a key and its corresponding value.
    
-   **Unique keys:** Duplicate keys are not allowed; assigning an existing key replaces its value.
    
-   **Allows duplicate values:** Multiple keys can have the same value.
    
-   **Keys must be hashable:** Strings, numbers, and suitable tuples can be keys; lists and dictionaries cannot.
    
-   **Values can be any data type:** Including lists, tuples, sets, dictionaries, and custom objects.
    
-   **Accessed using keys:** Values are retrieved using their corresponding keys, not numerical indexes.
    
-   **No positional indexing:** `my_dict[0]` looks for the key `0`, not the first entry.
    
-   **Iterable:** Can be traversed using loops.
    
-   **Dynamic size:** Can grow or shrink.
    
-   **Nested dictionaries:** Can contain other dictionaries and collections.
    
-   **Fast key lookup:** Accessing values by key is generally very efficient.

## **5\. Frozenset (**`frozenset`**)**

Example: `my_frozen = frozenset([10, 20, 30])`

-   **Unordered:** Does not maintain a guaranteed insertion order.
    
-   **Immutable:** Elements cannot be added or removed after creation.
    
-   **Does not allow duplicates:** Stores only unique elements.
    
-   **Allows mixed data types:** All elements must be hashable.
    
-   **Not indexable:** Cannot access elements using indexes.
    
-   **Iterable:** Can be traversed using loops.
    
-   **Supports mathematical set operations:** Union, intersection, difference, and symmetric difference.
    
-   **Hashable:** Can be used as a dictionary key or as an element of another set.
    
-   **Fixed size:** Cannot change after creation.
